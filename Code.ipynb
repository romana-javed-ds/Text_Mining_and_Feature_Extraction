```python
from pypdf import PdfReader

# Read teacher's stopwords PDF
reader = PdfReader("/content/Romana Javed - stopwords.pdf")

stopwords_text = ""

for page in reader.pages:
    stopwords_text += page.extract_text()

# Convert into lowercase words
stopwords = set(stopwords_text.lower().split())

print("Stopwords loaded successfully!")
print("Total stopwords:", len(stopwords))

print("\nSample stopwords:")
print(sorted(stopwords)[:30])
```

```python
import re

# Function for text preprocessing
def preprocess_text(text, language):

    # Convert text into lowercase
    text = str(text).lower()

    # Remove punctuation and special symbols
    text = re.sub(r'[^\w\s]', ' ', text)

    # Tokenize text into individual words
    tokens = re.findall(r'\b\w+\b', text)

    # Remove stopwords only from English documents
    if language == 'en':
        tokens = [word for word in tokens if word not in stopwords]

    return tokens


print("Preprocessing function created successfully!")
```

```python
# Apply preprocessing to all 50,000 records

df['tokens'] = df.apply(
    lambda row: preprocess_text(
        row['content'],
        row['language']
    ),
    axis=1
)

# Create cleaned text column
df['cleaned_content'] = df['tokens'].apply(
    lambda tokens: ' '.join(tokens)
)

print("Preprocessing completed successfully!")

print("Total documents processed:", len(df))
```

```python
# Select first 10 English documents

sample_docs = df[df['language'] == 'en'].head(10).copy()

print("Selected English Documents:", len(sample_docs))
```

```python
from nltk import pos_tag
from nltk.corpus import wordnet

# Convert POS tags into WordNet format
def get_wordnet_pos(tag):

    if tag.startswith('J'):
        return wordnet.ADJ

    elif tag.startswith('V'):
        return wordnet.VERB

    elif tag.startswith('N'):
        return wordnet.NOUN

    elif tag.startswith('R'):
        return wordnet.ADV

    return wordnet.NOUN


# Compare both techniques on 10 documents

results = []

for index, row in sample_docs.iterrows():

    tokens = row['tokens']

    # Apply Porter Stemming
    stemmed_words = [stemmer.stem(word) for word in tokens]

    # Apply POS-aware Lemmatization
    tagged_words = pos_tag(tokens)

    lemmatized_words = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_words
    ]

    results.append({
        'Document_ID': row['id'],
        'Original_Tokens': ' '.join(tokens[:30]),
        'Stemmed_Tokens': ' '.join(stemmed_words[:30]),
        'Lemmatized_Tokens': ' '.join(lemmatized_words[:30])
    })


comparison_df = pd.DataFrame(results)

display(comparison_df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

print("Bag of Words library imported successfully!")
```

```python
# Create Bag of Words model

vectorizer = CountVectorizer()

# Convert cleaned documents into text matrix
bow_matrix = vectorizer.fit_transform(
    df['cleaned_content'].fillna('')
)

print("Bag of Words created successfully!")

print("Matrix Shape:", bow_matrix.shape)

print("Total Documents:", bow_matrix.shape[0])

print("Total Unique Words:", bow_matrix.shape[1])
```

```python
import pandas as pd

# Get vocabulary
words = vectorizer.get_feature_names_out()

# Calculate total word frequency
word_counts = bow_matrix.sum(axis=0).A1

# Create frequency table
word_frequency = pd.DataFrame({
    'Word': words,
    'Frequency': word_counts
})

# Sort by frequency
word_frequency = word_frequency.sort_values(
    by='Frequency',
    ascending=False
)

# Display top 50 words
display(word_frequency.head(50))
```

```python
document_frequency = (bow_matrix > 0).sum(axis=0).A1

df_table = pd.DataFrame({
    'Word': words,
    'Document_Frequency': document_frequency
})

df_table = df_table.sort_values(
    by='Document_Frequency',
    ascending=False
)

display(df_table.head(50))
```

```python
word_frequency.head(50).to_csv(
    '/content/top_50_bow_words.csv',
    index=False
)

df_table.head(50).to_csv(
    '/content/bow_document_frequency.csv',
    index=False
)

print("Bag of Words results saved successfully!")
```

```python
# Create TF-IDF model

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    df['cleaned_content'].fillna('')
)

print("TF-IDF generated successfully!")

print("Number of Documents:", tfidf_matrix.shape[0])

print("Number of Features:", tfidf_matrix.shape[1])

print("Matrix Shape:", tfidf_matrix.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np
import pandas as pd

# Create TF-IDF model with vocabulary pruning
tfidf_vectorizer = TfidfVectorizer(
    min_df=5,
    max_df=0.85
)

tfidf_matrix = tfidf_vectorizer.fit_transform(
    df['cleaned_content'].fillna('')
)

print("TF-IDF matrix created!")
print("Matrix Shape:", tfidf_matrix.shape)
print("Documents:", tfidf_matrix.shape[0])
print("Vocabulary Size:", tfidf_matrix.shape[1])
```

```python
# Calculate aggregate TF-IDF scores
feature_names = tfidf_vectorizer.get_feature_names_out()

aggregate_scores = np.asarray(
    tfidf_matrix.sum(axis=0)
).ravel()

tfidf_ranking = pd.DataFrame({
    'Word': feature_names,
    'Aggregate_TFIDF': aggregate_scores
})

tfidf_ranking = tfidf_ranking.sort_values(
    by='Aggregate_TFIDF',
    ascending=False
).reset_index(drop=True)

print("Top 20 TF-IDF Terms:")
display(tfidf_ranking.head(20))
```

```python
# Select different values of K
K_values = [50, 100, 200]

for K in K_values:

    actual_K = min(K, len(tfidf_ranking))

    # Select top K words
    top_words = tfidf_ranking.head(actual_K)['Word'].tolist()

    # Get their column positions
    feature_indices = [
        list(feature_names).index(word)
        for word in top_words
    ]

    # Extract selected TF-IDF vectors
    selected_matrix = tfidf_matrix[:, feature_indices]

    # Save vectors as NumPy file
    np.save(
        f'/content/tfidf_top_{K}_vectors.npy',
        selected_matrix.toarray().astype('float32')
    )

    # Save corresponding word rankings
    tfidf_ranking.head(actual_K).to_csv(
        f'/content/tfidf_top_{K}_words.csv',
        index=False
    )

    print(f"K = {K}")
    print("Vector Shape:", selected_matrix.shape)
    print("Saved successfully!")
    print("-" * 40)
```

```python
tfidf_ranking.to_csv(
    '/content/complete_tfidf_ranking.csv',
    index=False
)

print("Complete TF-IDF ranking saved!")
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np

# Select Top 100 TF-IDF features
top_100_words = tfidf_ranking.head(100)['Word'].tolist()

feature_indices = [
    np.where(feature_names == word)[0][0]
    for word in top_100_words
]

X_top100 = tfidf_matrix[:, feature_indices].toarray()

# Apply PCA to reduce dimensions to 2
pca = PCA(n_components=2, random_state=42)

pca_results = pca.fit_transform(X_top100)

# Create PCA DataFrame
pca_df = pd.DataFrame({
    'PC1': pca_results[:, 0],
    'PC2': pca_results[:, 1],
    'HTTP_Status': df['http_status'].values
})

# Plot PCA visualization
plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    c=pca_df['HTTP_Status'],
    cmap='viridis',
    alpha=0.5,
    s=10
)

plt.colorbar(scatter, label='HTTP Status')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Visualization of Top 100 TF-IDF Features')

plt.tight_layout()
plt.savefig('/content/PCA_visualization.png', dpi=300)
plt.show()

print("PCA Visualization completed!")
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Explained Variance:",
      pca.explained_variance_ratio_.sum())
```
